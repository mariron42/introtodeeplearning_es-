## 1.1 ¿Qué es PyTorch?

PyTorch es una biblioteca de aprendizaje automático (machine learning), como TensorFlow. En su núcleo, PyTorch proporciona una interfaz para crear y manipular [tensores](https://pytorch.org/docs/stable/tensors.html), que son estructuras de datos que puedes considerar como arreglos multidimensionales. Los tensores se representan como arreglos n-dimensionales de tipos de datos base como cadenas de texto o enteros -- proporcionan una forma de generalizar vectores y matrices a dimensiones superiores. PyTorch proporciona la capacidad de realizar cómputos sobre estos tensores, definir redes neuronales y entrenarlas de manera eficiente.

La propiedad [```shape```](https://pytorch.org/docs/stable/generated/torch.Tensor.shape.html#torch.Tensor.shape) de un tensor de PyTorch define su número de dimensiones y el tamaño de cada dimensión. El `ndim` o [```dim```](https://pytorch.org/docs/stable/generated/torch.Tensor.dim.html#torch.Tensor.dim) de un tensor de PyTorch proporciona el número de dimensiones (n-dimensiones) -- esto es equivalente al rango del tensor (como se usa en TensorFlow), y también puedes pensar en esto como el orden o grado del tensor.

Comencemos creando algunos tensores e inspeccionando sus propiedades:


````python
integer = torch.tensor(1234)
decimal = torch.tensor(3.14159265359)

print(f"`integer` is a {integer.ndim}-d Tensor: {integer}")
print(f"`decimal` is a {decimal.ndim}-d Tensor: {decimal}")
````

## 1.4 Diferenciación Automática en PyTorch

En PyTorch, [`torch.autograd`](https://pytorch.org/docs/stable/autograd.html) se utiliza para la [diferenciación automática](https://en.wikipedia.org/wiki/Automatic_differentiation), que es fundamental para entrenar modelos de aprendizaje profundo con [retropropagación](https://en.wikipedia.org/wiki/Backpropagation) (backpropagation).

Usaremos el método [`.backward()`](https://pytorch.org/docs/stable/generated/torch.Tensor.backward.html) de PyTorch para rastrear operaciones y calcular gradientes. En un tensor, el atributo [`requires_grad`](https://pytorch.org/docs/stable/generated/torch.Tensor.requires_grad_.html) controla si autograd debe registrar las operaciones sobre ese tensor. Cuando se realiza un paso hacia adelante (forward pass) a través de la red, PyTorch construye un grafo computacional dinámicamente; luego, para calcular el gradiente, se llama al método `backward()` para realizar la retropropagación.

Calculemos el gradiente de $ y = x^2 $:

````python
### Cálculo de gradientes ###

# y = x^2
# Ejemplo: x = 3.0
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2
y.backward()  # Calcular el gradiente

dy_dx = x.grad
print("dy_dx of y=x^2 at x=3.0 is: ", dy_dx)
assert dy_dx == 6.0
````
