## 1.1 ¿Qué es PyTorch?

PyTorch es una biblioteca de aprendizaje automático, como TensorFlow. En su núcleo, PyTorch proporciona una interfaz para crear y manipular [tensores](https://pytorch.org/docs/stable/tensores.html), que son estructuras de datos que puedes considerar como arreglos multidimensionales. Los tensores se representan como arreglos n-dimensionales de tipos de datos base como cadenas o enteros -- proporcionan una manera de generalizar vectores y matrices a dimensiones superiores. PyTorch proporciona la capacidad de realizar cálculos en estos tensores, definir redes neuronales y entrenarlas eficientemente.

The [```shape```](https://pytorch.org/docs/stable/generated/torch.Tensor.shape.html#torch.Tensor.shape) of a PyTorch tensor define su número de dimensiones y el tamaño de cada dimensión. The `ndim` or [```dim```](https://pytorch.org/docs/stable/generated/torch.Tensor.dim.html#torch.Tensor.dim) of a PyTorch tensor proporciona el número de elementos (n-dimensions) -- esto es equivalente al rango del tensor (as is used in TensorFlow), y también puedes pensar en esto como el orden o grado del tensor.

Let’s start by creating some tensores y inspecting their properties:


````python
integer = torch.tensor(1234)
decimal = torch.tensor(3.14159265359)

print(f"`integer` is a {integer.ndim}-d Tensor: {integer}")
print(f"`decimal` is a {decimal.ndim}-d Tensor: {decimal}")
````

## 1.4 Diferenciación Automática en PyTorch

En PyTorch, [`torch.autograd`](https://pytorch.org/docs/stable/autograd.html) se usa para [automatic differentiation](https://en.wikipedia.org/wiki/Automatic_differentiation), que es crítica para entrenar modelos de aprendizaje profundo con [backpropagation](https://en.wikipedia.org/wiki/Backpropagation).

Usaremos the PyTorch [`.backward()`](https://pytorch.org/docs/stable/generated/torch.Tensor.backward.html) method to trace operations for computing gradients. On a tensor, the [`requires_grad`](https://pytorch.org/docs/stable/generated/torch.Tensor.requires_grad_.html) attribute controls whether autograd should record operations on that tensor. When a forward pass is made through the network, PyTorch builds a computational graph dynamically; then, to compute the gradient, the `backward()` method is called to perform backpropagation.

Calculemos el gradiente de $ y = x^2 $:

````python
# ## Gradient computation ###

# y = x^2
# Ejemplo: x = 3.0
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2
y.backward()  # Compute the gradient

dy_dx = x.grad
print("dy_dx of y=x^2 at x=3.0 is: ", dy_dx)
assert dy_dx == 6.0
````

En el entrenamiento de redes neuronales, usamos diferenciación y descenso de gradiente estocástico (SGD) para optimizar una función de pérdida.
