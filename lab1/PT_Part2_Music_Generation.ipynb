## 2.4 El modelo de Red Neuronal Recurrente (RNN)

Ahora estamos listos para definir y entrenar un modelo RNN en nuestro conjunto de datos de música ABC, y luego usar ese modelo entrenado para generar una nueva canción. Entrenaremos nuestra RNN usando lotes de fragmentos de canciones de nuestro conjunto de datos, que generamos en la sección anterior.

El modelo se basa en la arquitectura LSTM, donde usamos un vector de estado para mantener información sobre las relaciones temporales entre caracteres consecutivos. La salida final del LSTM se alimenta luego a una capa lineal completamente conectada [`nn.Linear`](https://pytorch.org/docs/stable/generated/torch.nn.Linear.html) donde produciremos un softmax sobre cada carácter del vocabulario, y luego muestrearemos de esta distribución para predecir el siguiente carácter.

Como introdujimos en la primera parte de este laboratorio, usaremos [`nn.Module`](https://pytorch.org/docs/stable/generated/torch.nn.Module.html) de PyTorch para definir el modelo. Se utilizan tres componentes para definir el modelo:

* [`nn.Embedding`](https://pytorch.org/docs/stable/generated/torch.nn.Embedding.html): Esta es la capa de entrada, que consiste en una tabla de búsqueda entrenable que mapea los números de cada carácter a un vector con dimensiones `embedding_dim`.
* [`nn.LSTM`](https://pytorch.org/docs/stable/generated/torch.nn.LSTM.html): Nuestra red LSTM, con tamaño `hidden_size`.
* [`nn.Linear`](https://pytorch.org/docs/stable/generated/torch.nn.Linear.html): La capa de salida, con `vocab_size` salidas.

### Definir el modelo RNN

Definamos nuestro modelo como un `nn.Module`. Completa los `TODOs` para definir el modelo RNN.


```python
### Definición del modelo RNN ###

'''POR HACER: Añadir capas LSTM y Linear para definir el modelo RNN usando nn.Module'''
class LSTMModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_size):
        super(LSTMModel, self).__init__()
        self.hidden_size = hidden_size

        # Definir cada una de las capas de la red
        # Capa 1: Capa de embedding para transformar índices en vectores densos
        #   de un tamaño de embedding fijo
        self.embedding = nn.Embedding(vocab_size, embedding_dim)

        '''POR HACER: Capa 2: LSTM con hidden_size `hidden_size`. nota: el número de capas por defecto es 1.
         Usa el módulo nn.LSTM() de pytorch.'''
        self.lstm = nn.LSTM('''POR HACER''') # POR HACER

        '''POR HACER: Capa 3: Capa Linear (completamente conectada) que transforma la salida del LSTM
        #   al tamaño del vocabulario.'''
        self.fc = nn.Linear('''POR HACER''') # POR HACER

    def init_hidden(self, batch_size, device):
        # Inicializar el estado oculto y el estado de celda con ceros
        return (torch.zeros(1, batch_size, self.hidden_size).to(device),
                torch.zeros(1, batch_size, self.hidden_size).to(device))

    def forward(self, x, state=None, return_state=False):
        x = self.embedding(x)

        if state is None:
            state = self.init_hidden(x.size(0), x.device)
        out, state = self.lstm(x, state)

        out = self.fc(out)
        return out if not return_state else (out, state)


```

## 2.6 Generar música usando el modelo RNN

Ahora, ¡podemos usar nuestro modelo RNN entrenado para generar algo de música! Al generar música, tendremos que alimentar al modelo algún tipo de semilla para que comience (¡porque no puede predecir nada sin algo con qué empezar!).

Una vez que tengamos una semilla generada, podemos predecir iterativamente cada carácter sucesivo (recuerda, estamos usando la representación ABC para nuestra música) usando nuestra RNN entrenada. Más específicamente, recuerda que nuestra RNN produce un `softmax` sobre posibles caracteres sucesivos. Para la inferencia, muestreamos iterativamente de estas distribuciones, y luego usamos nuestras muestras para codificar una canción generada en formato ABC.

¡Entonces, todo lo que tenemos que hacer es escribirlo en un archivo y escucharlo!

### El procedimiento de predicción

Ahora, estamos listos para escribir el código que genera texto en formato de música ABC:

* Inicializar una cadena de inicio "semilla" y el estado de la RNN, y establecer el número de caracteres que queremos generar.

* Usar la cadena de inicio y el estado de la RNN para obtener la distribución de probabilidad sobre el siguiente carácter predicho.

* Muestrear de una distribución multinomial para calcular el índice del carácter predicho. Este carácter predicho se usa luego como la siguiente entrada al modelo.

* En cada paso de tiempo, el estado actualizado de la RNN se retroalimenta al modelo, de modo que ahora tiene más contexto para hacer la siguiente predicción. Después de predecir el siguiente carácter, los estados actualizados de la RNN se retroalimentan nuevamente al modelo, que es como aprende las dependencias de secuencia en los datos, ya que obtiene más información de las predicciones anteriores.
