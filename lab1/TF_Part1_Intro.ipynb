## 1.1 ¿Por qué TensorFlow se llama TensorFlow?

TensorFlow se llama 'TensorFlow' porque maneja el flujo (nodo/operación matemática) de Tensores, que son estructuras de datos que puedes considerar como arreglos multidimensionales. Los Tensores se representan como arreglos n-dimensionales de tipos de datos básicos como cadenas de texto o enteros -- proporcionan una forma de generalizar vectores y matrices a dimensiones superiores.

La ```shape``` (forma) de un Tensor define su número de dimensiones y el tamaño de cada dimensión. El ```rank``` (rango) de un Tensor proporciona el número de dimensiones (n-dimensiones) -- también puedes pensar en esto como el orden o grado del Tensor.

Primero veamos los Tensores de dimensión 0, de los cuales un escalar es un ejemplo:



````python
sport = tf.constant("Tennis", tf.string)
number = tf.constant(1.41421356237, tf.float64)

print("`sport` is a {}-d Tensor".format(tf.rank(sport).numpy()))
print("`number` is a {}-d Tensor".format(tf.rank(number).numpy()))
````

Se pueden usar vectores y listas para crear Tensores de 1 dimensión:



````python
sports = tf.constant(["Tennis", "Basketball"], tf.string)
numbers = tf.constant([3.141592, 1.414213, 2.71821], tf.float64)

print("`sports` is a {}-d Tensor with shape: {}".format(tf.rank(sports).numpy(), tf.shape(sports)))
print("`numbers` is a {}-d Tensor with shape: {}".format(tf.rank(numbers).numpy(), tf.shape(numbers)))
````

## 1.4 Diferenciación automática en TensorFlow

La [diferenciación automática](https://en.wikipedia.org/wiki/Automatic_differentiation)
es una de las partes más importantes de TensorFlow y es la columna vertebral del entrenamiento con
[retropropagación](https://en.wikipedia.org/wiki/Backpropagation). Usaremos el GradientTape de TensorFlow [`tf.GradientTape`](https://www.tensorflow.org/api_docs/python/tf/GradientTape?version=stable) para rastrear operaciones y calcular gradientes posteriormente.

Cuando se realiza un paso hacia adelante (forward pass) a través de la red, todas las operaciones del paso hacia adelante se graban en una "cinta" (tape); luego, para calcular el gradiente, la cinta se reproduce hacia atrás. Por defecto, la cinta se descarta después de reproducirse hacia atrás; esto significa que un `tf.GradientTape` particular solo puede
calcular un gradiente, y las llamadas posteriores generan un error en tiempo de ejecución. Sin embargo, podemos calcular múltiples gradientes sobre el mismo cálculo creando una cinta de gradientes ```persistent``` (persistente).

Primero, veremos cómo podemos calcular gradientes usando GradientTape y acceder a ellos para el cálculo. Definimos la función simple $ y = x^2$ y calculamos el gradiente:



````python
### Cálculo de gradientes con GradientTape ###

# y = x^2
# Ejemplo: x = 3.0
x = tf.Variable(3.0)

# Iniciar la cinta de gradientes
with tf.GradientTape() as tape:
  # Definir la función
  y = x * x
# Acceder al gradiente -- derivada de y con respecto a x
dy_dx = tape.gradient(y, x)

assert dy_dx.numpy() == 6.0
````

En el entrenamiento de redes neuronales, usamos diferenciación y descenso de gradiente estocástico (SGD) para optimizar una función de pérdida.
